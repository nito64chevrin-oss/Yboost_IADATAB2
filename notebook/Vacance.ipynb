```python
import pandas as pd
```

```python
vacances = pd.read_csv("../data/raw/vacances_scolaires.csv", parse_dates=["date"])

# Nombre de zones en vacances chaque jour (True compte pour 1, False pour 0)
vacances["nb_zones_vacances"] = (
    vacances[["vacances_zone_a", "vacances_zone_b", "vacances_zone_c"]].sum(axis=1)
)

print("Premier jour :", vacances["date"].min())
print("Dernier jour :", vacances["date"].max())
vacances[vacances["date"].between("2024-02-08", "2024-02-20")]
```

```python
from jours_feries_france import JoursFeries

# Liste de tous les jours fériés de 1993 à 2025
dates_feries = []
for annee in range(1993, 2026):
    dates_feries.extend(JoursFeries.for_year(annee).values())

feries = pd.DataFrame({"date": pd.to_datetime(dates_feries), "ferie": 1})
feries.head(11)
```

```python
jours = pd.DataFrame({"date": pd.date_range("1993-01-01", "2025-12-31", freq="D")})

jours = jours.merge(vacances[["date", "nb_zones_vacances"]], on="date", how="left")
jours = jours.merge(feries, on="date", how="left")

# Un jour absent de la liste des fériés n'est pas férié
jours["ferie"] = jours["ferie"].fillna(0).astype(int)

jours.head()
```
